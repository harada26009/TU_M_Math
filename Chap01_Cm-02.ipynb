```
def prime_N(n):
    if n < 2:
        return[]
    primeN = []
    for num in range(2, n+1):
        prime = True
        limit = int(num**0.5) +1
        for i in range(2, limit):
           if num % i ==0:
              prime = False
              break
        if prime:
           primeN.append(num)
    return primeN

try:
    inputprime = input("数字を入力してください :")
    limit = int(inputprime)
    result = prime_N(limit)

    print(f"{limit}までの素数は以下の通りです :")
    print(result)
    print(f"合計{len(result)}個見つかりました。")

except ValueError:
    print("エラー:数字を入力してください")
```
