```python
from ast import increment_lineno
import numpy as np
import numpy.random as random
import scipy as sp

import matplotlib.pyplot as plot
import matplotlib as mpl
%matplotlib inline

%precision 3
```

```python
def f(x):
  y = 5*x -10
  return y

x = np.linspace(0,4)
plot.plot(x,f(x))
plot.plot(x,np.zeros(len(x)))
plot.grid(True)
```

```python
from scipy.optimize import fsolve

x = fsolve(f,2)
print(x)
```
