```python
from sklearn.linear_model import LinearRegression, Lasso

X = auto.drop('price',axis=1)
y = auto['price']
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.5,random_state=0)

models = {
    'linear':LinearRegression(),
    'lasso1':Lasso(alpha=1.0,random_state=0),
    'lasso2':Lasso(alpha=200.0,random_state=0)
}

scores = {}
for model_name,model in models.items():
    model.fit(X_train,y_train)
    scores[(model_name,'train')] = model.score(X_train,y_train)
    scores[(model_name,'test')] = model.score(X_test,y_test)

pd.Series(scores).unstack()
```

ラッソ回帰とリッジ回帰に用いられる正則化項の目的とメリット

ラッソ回帰（Lasso Regression）とリッジ回帰（Ridge Regression）は、どちらも正則化を用いて回帰モデルの過学習を防ぐ手法である。正則化とは、回帰係数が大きくなりすぎないように制約（ペナルティ）を加えることで、モデルの複雑さを抑え、未知のデータに対する予測性能を向上させることを目的としている。

リッジ回帰では、回帰係数の二乗和（L2正則化）を目的関数に加える。これにより、回帰係数を全体的に小さくすることができるが、通常は係数を0にはしない。そのため、すべての説明変数を利用しながら過学習を抑制できる。また、説明変数同士の相関が強い多重共線性が存在する場合でも、係数のばらつきを抑え、安定した推定結果を得られるというメリットがある。

一方、ラッソ回帰では、回帰係数の絶対値の和（L1正則化）を目的関数に加える。L1正則化では、一部の回帰係数を0にする性質があるため、不要な説明変数を自動的に除外できる。つまり、パラメーター（特徴量）選択を同時に行えることが大きな特徴である。説明変数が非常に多い場合や、重要な特徴量だけを残したい場合に有効である。

両者とも過学習を防ぎ、未知データへの予測性能を向上させることを目的としているが、特徴が異なる。リッジ回帰は多重共線性に強く、すべての説明変数を利用したまま安定したモデルを構築したい場合に適している。一方、ラッソ回帰は不要な説明変数を削除し、モデルを簡潔にできるため、特徴量が多いデータや変数選択を行いたい場合に適している。

このように、リッジ回帰とラッソ回帰はどちらも正則化によってモデルの性能向上を目指す手法であるが、多重共線性への対策を重視するならリッジ回帰、特徴量選択を重視するならラッソ回帰を選択するのが一般的である。
