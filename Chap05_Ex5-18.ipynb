```python
from ast import increment_lineno
import numpy as np
import numpy.random as random
import scipy as sp

import matplotlib.pyplot as plot
import matplotlib as mpl
%matplotlib inline

%precision 3
```

```python
def f2(x):
  y = x**3 -2 * x**2 -11 * x +12
  return y

x = np.linspace(-5,5)
plot.plot(x,f2(x))
plot.plot(x,np.zeros(len(x)))
plot.grid(True)
```

```python
from scipy.optimize import fsolve

x = fsolve(f2,-3)
print(x)
```

```python
x = fsolve(f2,1)
print(x)
```

```python
x = fsolve(f2,4)
print(x)
```
